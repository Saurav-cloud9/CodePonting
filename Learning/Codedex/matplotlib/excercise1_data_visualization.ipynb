```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
year = np.array([2018, 2019, 2020, 2021, 2022, 2023, 2024, 2025])
revenue = np.array([4.8, 5.3, 2.1, 3.4, 5.9, 6.5, 7.2, 8.1])

plt.figure()
plt.plot(year, revenue)

```

```python
year = np.array([2012, 2013, 2014, 2015, 2016, 2017, 2018, 2019, 2022, 2023])
temperature = np.array([30, 40, 35, 25, 42, 26, 33, 37, 29, 34])

plt.Figure()
plt.plot(year, temperature)

plt.title('year and temp')
plt.xlabel('year')
plt.ylabel('temperature')

plt.show()
```
