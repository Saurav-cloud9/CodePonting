Go into your phone and find the number of steps you’ve taken in the last week and add them into an array.

Find the min, max, sum, and average using .min(), .max(), .sum(), .average().

```python
import numpy as np
```

```python
temperatures = np.array([[50, 51, 54, 59, 59, 53, 54, 51],
                         [45, 50, 38, 44, 40, 46, 43, 39]])
```

```python
print(np.sum(temperatures, axis=0))   # Sum of each column
print(np.min(temperatures, axis=1))   # Min of each row
```

```python
steps_taken = np.array([[3000, 2000, 4500, 2300, 1900, 2100, 4200], 
                        [3300, 3800, 4200, 3600, 2750, 3150, 3000]])
```

```python
print(np.min(steps_taken, axis=1))
print(np.max(steps_taken, axis=1))
print(np.sum(steps_taken, axis=1))
print(np.round(np.average(steps_taken, axis=1), 2))
```
