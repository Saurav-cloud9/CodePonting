```python
import pandas as pd
import mplfinance as mpf

# Load data
df = pd.read_parquet(r'C:\Users\saurav\CodePonting\Algo_Trading\Framework_V1\data\historical\daily\NIFTY50.parquet')

# Convert datetime column
df['datetime'] = pd.to_datetime(df['datetime'], unit='ms')
df.set_index('datetime', inplace=True)

# Rename columns for mplfinance (requires: Open, High, Low, Close, Volume)
df.rename(columns={'open': 'Open', 'high': 'High', 'low': 'Low', 'close': 'Close', 'volume': 'Volume'}, inplace=True)

# Plot last 100 candles
mpf.plot(df.tail(100), type='candle', style='charles', volume=True, title='NIFTY50 - Last 100 Days')
```
