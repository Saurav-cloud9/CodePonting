```python
import pandas as pd
from pathlib import Path
import pyarrow.parquet as pq

# ---- 1️⃣ Create 10 test candles (TATASTEEL-style) ----
data = {
    "datetime": pd.date_range("2024-01-01 09:15", periods=10, freq="5min"),
    "open":  [136.4,136.2,135.85,136.15,135.8,136,136.05,136.25,136.4,136.5],
    "high":  [136.4,136.2,136.15,136.15,136.2,136.15,136.3,136.55,136.55,137.2],
    "low":   [135.45,135.75,135.55,135.7,135.8,135.85,136,136.25,136.3,136.5],
    "close": [136.2,135.85,136.15,135.85,136,136.05,136.3,136.4,136.55,137.1],
    "volume":[974407,632912,534765,394908,459369,441722,268912,705535,317530,1173917],
    "oi":    [0]*10
}

df_test = pd.DataFrame(data)

# ---- 2️⃣ Save paths ----
base = Path(r"C:\Users\saurav\CodePonting\Algo_Trading\Framework_V1\data\historical\intraday_5min")
base.mkdir(parents=True, exist_ok=True)

parquet_path = base / "TATASTEEL_10.parquet"
csv_path = base / "TATASTEEL_10.csv"

# ---- 3️⃣ Save files ----
df_test.to_parquet(parquet_path, index=False)
df_test.to_csv(csv_path, index=False)

print("Saved:")
print(parquet_path)
print(csv_path)

# ---- 4️⃣ Show pandas dtype ----
print("\nPandas dtypes:")
print(df_test.dtypes)

# ---- 5️⃣ Inspect true parquet schema ----
table = pq.read_table(parquet_path)
print("\nParquet schema:")
print(table.schema)

# ---- 6️⃣ Reload preview ----
df_reload = pd.read_parquet(parquet_path)
display(df_reload)


```
