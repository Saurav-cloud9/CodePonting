```python
import pandas as pd
import yfinance as yf

# Download data (creates multi-level columns)
df = yf.download("^NSEI", start="2024-01-01", end="2024-01-10", progress=False)

print("=" * 70)
print("BEFORE FLATTENING:")
print("=" * 70)
print(f"Column structure: {df.columns}")
print(f"Column type: {type(df.columns)}")
print()
print(type(df['High']))

# Try to access 'High'
high = df['High']
print(f"high = df['High']")
print(f"Type: {type(high)}")
print(f"Shape: {high.shape}")
print(f"Data:\n{high.head()}")
print()

# Try boolean indexing (this will FAIL)
try:
    mask = high > 23000
    print(f"mask type: {type(mask)}")
    print(f"mask shape: {mask.shape}")

    test_series = pd.Series([1, 2, 3, 4, 5], index=df.index[:5])
    result = test_series[mask[:5]]  # This will error
except Exception as e:
    print(f"❌ ERROR with multi-level: {type(e).__name__}: {e}")

print("\n" + "=" * 70)
print("AFTER FLATTENING:")
print("=" * 70)
print(type(df['High']))

# Flatten columns
df.columns = df.columns.get_level_values(0)
print(f"Column structure: {df.columns}")
print(f"Column type: {type(df.columns)}")
print()

# Now access 'High'
high = df['High']
print(f"high = df['High']")
print(f"Type: {type(high)}")
print(f"Shape: {high.shape}")
print(f"Data:\n{high.head()}")
print()

# Boolean indexing now WORKS
mask = high > 23000
print(f"mask type: {type(mask)}")
print(f"mask shape: {mask.shape}")

test_series = pd.Series([1, 2, 3, 4, 5], index=df.index[:5])
result = test_series[mask[:5]]
print(f"✅ Indexing works! Result: {result.values}")
```
